```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
```

```python
d={"col_1" : [20,30,10,np.nan,50],
   "col_2" : [100,np.nan,300,400,500],
   "col_3":['B','A','C','D',np.nan]}
```

```python
df=pd.DataFrame(d)
```

```python
impute1=SimpleImputer(strategy='constant',fill_value=0)
```

```python
impute2=SimpleImputer(strategy='constant',fill_value='value required')
```

```python
df
```

```python
impute1.fit_transform(df[['col_1','col_2']])
#it accpct only 2d array
```

```python
impute2.fit_transform(df[['col_3']])
```

```python
impute_mean=SimpleImputer(strategy='mean',fill_value='')
```

```python
impute_mean.fit_transform(df[['col_1']])
```

```python
impute_median=SimpleImputer(strategy='median')
impute_median.fit_transform(df[['col_2','col_1']])

# #if we use median, only we can use in numerical col we cann't use in categorical col
# '''impute_median.fit_transform(df[['col_2','col_3']])
# ValueError: Cannot use median strategy with non-numeric data:'''
```

```python
impute_mf=SimpleImputer(strategy='most_frequent')
impute_mf.fit_transform(df[['col_1','col_3']])
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix,precision_score,recall_score,accuracy_score,f1_score
import seaborn as sns
from sklearn.preprocessing import StandardScaler,MinMaxScaler,RobustScaler
import matplotlib.pyplot as plt
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,OrdinalEncoder
from sklearn.metrics import classification_report
```

```python
df=pd.read_csv('loan_data.csv')
df
```

```python
df.info()
```

```python
df.head()
```

```python
x=df.drop(columns='loan_status')
y=df['loan_status']
```

```python
xtrain,xtest,ytrian,ytest=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
xtrain
```

```python
xtest
```

```python
cat_col=x.select_dtypes(include='object').columns
num_col=x.select_dtypes(include='number').columns
```

```python
cat_col
```

```python
num_col
```

```python
plt.figure(figsize=(15,7))
sns.boxplot(df[num_col])
```

```python
num_col
```

```python
preprocessing=ColumnTransformer(transformers=[
                                              ('rob_scaler',RobustScaler(),num_col),
                                              ('ohe_encoder',OneHotEncoder(handle_unknown='ignore'),cat_col),
                                              ('odr_encoder',OrdinalEncoder(handle_unknown='error'),['person_education'])
                                              ]
)
preprocessing.fit(xtrain,ytrian)
```

```python
main_pipeline=Pipeline(
                    steps=[
                           ('per',preprocessing),
                           ('model',LogisticRegression())
                    ]
)
main_pipeline.fit(xtrain,ytrian)
```

```python
main_pipeline
```

*model evaluation*

# Classification Metrics

1-->cnfusion matrix

```python
ytrian.value_counts()
```

```python
ytrain_pred=main_pipeline.predict(xtrain)
cm_train=confusion_matrix(ytrian,ytrain_pred)
cm_train
```

```python
ytest_pred=main_pipeline.predict(xtest)
cm_test=confusion_matrix(ytest,ytest_pred)
cm_test
```

```python
print(classification_report(ytrian,ytrain_pred))
#support---> number of lables
# macro--> when we have imbalance dataset we use it,(ex-->0.93+0.74 divided By 2)
# weighted avg-->it check weight of each metrics 
```

2---> Precison_score for 1

```python
ytrain_pred=main_pipeline.predict(xtrain)
cm_train=precision_score(ytrian,ytrain_pred,pos_label=1)
cm_train
```

```python
ytest_pred=main_pipeline.predict(xtest)
cm_test=precision_score(ytest,ytest_pred,pos_label=1)
cm_test
```

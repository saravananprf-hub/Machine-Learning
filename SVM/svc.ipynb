```python
import pandas as pd
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.svm import SVC
```

```python
df=pd.read_csv('loan_data.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
x=df.drop(columns='loan_status')
y=df['loan_status']
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
cat_cols=x.select_dtypes(include='object').columns
num_cols=x.select_dtypes(include='number').columns
```

```python
preprocessing=ColumnTransformer(transformers=[
                                              ('encoding',OneHotEncoder(handle_unknown='ignore'),cat_cols),
                                              ('scaling',StandardScaler(),num_cols)

])
```

```python
main_pipeline=Pipeline(steps=[
                              ('pre',preprocessing),
                              ('model',SVC(random_state=42,class_weight='balanced'))
])
grid_cv=GridSearchCV(estimator=main_pipeline,
                     param_grid={'model__C':[0.001,0.01,0.1,1,10],
                                 'model__kernel':['rbf','linear','poly','sigmoid'],
                                 'model__degree':[2,3,5]},
                                 verbose=1)
grid_cv.fit(xtrain,ytrain)
```

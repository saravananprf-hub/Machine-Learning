```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
```

```python
d={'product':['kiki','oooo','pppp','pppp','kiki','kiki','oooo','oooo','oooo','oooo'],
   'tpyes':['maggi','yeepi','maggi','maggi','maggi','yeepi','dosa','dosa','maggi','dosa'],
   'target':[100,200,300,400,500,600,600,700,800,900]}
```

```python
df=pd.DataFrame(d)
```

```python
x=df.drop(columns='target')
y=df['target']
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
xtrain
```

```python
enncoder=OneHotEncoder(sparse_output=False,handle_unknown='ignore')
enncoder.fit_transform(xtrain)
```
